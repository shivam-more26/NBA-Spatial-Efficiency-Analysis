Phase 2: Feature Engineering.This notebook reads the cleaned CSV produced in Phase 1 and performs light feature engineering to derive shot-value analytics. It creates two new mathematical columns from the existing raw fields, SHOT_TYPE and SHOT_MADE_FLAG, transforming categorical and boolean data into quantifiable scoring metrics ready for downstream analysis (efficiency, heatmaps, player comparisons).

```python
import pandas as pd
```

```python
file_path = "GSW_2022_Roster_Shots.csv"
try:
    df = pd.read_csv(file_path)
    print(f"Successfully loaded {len(df)} shots.")
except FileNotFoundError:
    print(f"ERROR: Could not find {file_path}. Make sure it is uploaded to your Colab session!")
    raise

print("Calculating Expected Value metrics...")
```

```python
# New Features
df['SHOT_VALUE'] = df['SHOT_TYPE'].apply(lambda x: 3 if '3PT' in str(x) else 2)
df['POINTS_SCORED'] = df['SHOT_MADE_FLAG'] * df['SHOT_VALUE']
```

```python
preview_cols = ['PLAYER_NAME', 'SHOT_TYPE', 'SHOT_MADE_FLAG', 'SHOT_VALUE', 'POINTS_SCORED']
print("\n--- Data Preview ---")
print(df[preview_cols].head(5))
```

```python
output_filename = "GSW_2022_Processed_EV.csv"
df.to_csv(output_filename, index=False)
print(f"\nSuccess! Processed data saved as: {output_filename}")
try:
    from google.colab import files
    files.download(output_filename)
except ImportError:
    print("Not running in Colab - file saved locally.")
```
