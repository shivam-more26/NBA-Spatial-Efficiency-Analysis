Phase 2: Feature Engineering.This notebook reads the cleaned CSV produced in Phase 1 and performs light feature engineering to derive shot-value analytics. It creates two new mathematical columns from the existing raw fields, SHOT_TYPE and SHOT_MADE_FLAG, transforming categorical and boolean data into quantifiable scoring metrics ready for downstream analysis (efficiency, heatmaps, player comparisons).

```python
import random
import numpy as np
import pandas as pd
from pathlib import Path
np.random.seed(42)
random.seed(42)
```

## Data

```python
repo_root = Path.cwd().resolve()
if repo_root.name == "notebooks":
    repo_root = repo_root.parent
raw_path = repo_root / "data" / "raw" / "gsw_2022_roster_shots.csv"
file_path = raw_path
try:
    df = pd.read_csv(file_path)
    print(f"Successfully loaded {len(df)} shots.")
except FileNotFoundError:
    print(f"ERROR: Could not find {file_path}. Make sure the project root is correct.")
    raise

print("Calculating Expected Value metrics...")
```

## Modeling

## Cleaning

```python
# New Features
df['SHOT_VALUE'] = df['SHOT_TYPE'].apply(lambda x: 3 if '3PT' in str(x) else 2)
df['POINTS_SCORED'] = df['SHOT_MADE_FLAG'] * df['SHOT_VALUE']
```

## EDA
